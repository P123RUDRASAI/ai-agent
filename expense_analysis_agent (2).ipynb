```python
import json
import os
import matplotlib.pyplot as plt
import pandas as pd
from google import genai
from google.genai import types

# Ensure SAMPLE_EXPENSES, client, and model are globally defined from previous cells.

def categorize_expenses(client_obj: genai.Client, expenses: list) -> pd.DataFrame:
    """Uses LLM structured output to categorize raw expense descriptions."""
    if not expenses:
        return pd.DataFrame(columns=["date", "description", "amount", "category"])

    prompt = f"""
    Categorize each expense into one of the following categories:
    [Food & Dining, Transportation, Utilities, Entertainment, Shopping, Miscellaneous]

    Input Data:
    {json.dumps(expenses)}

    Return a JSON array where each object contains:
    - 'date': string
    - 'description': string
    - 'amount': float
    - 'category': string
    """

    # Use the global 'model' variable
    response = client_obj.models.generate_content(
        model=model, # Using the global 'model' variable
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            temperature=0.1,
        ),
    )

    categorized_data = json.loads(response.text)
    return pd.DataFrame(categorized_data)


def plot_expense_breakdown(df: pd.DataFrame, output_path: str = "expense_summary.png") -> str:
    """Generates and saves a pie chart of expenses by category."""
    if df.empty or "category" not in df.columns:
        raise ValueError("DataFrame must contain categorized expense data.")

    category_totals = df.groupby("category")["amount"].sum()

    plt.figure(figsize=(7, 7))
    plt.pie(
        category_totals,
        labels=category_totals.index,
        autopct="%1.1f%%",
        startangle=140,
    )
    plt.title("Expense Breakdown by Category")
    plt.tight_layout()
    plt.savefig(output_path)
    plt.close()

    return output_path


def generate_financial_advice(client_obj: genai.Client, df: pd.DataFrame) -> str:
    """Generates personalized financial advice based on spending patterns."""
    if df.empty:
        return "No expenses to analyze."

    total_spent = df["amount"].sum()
    summary_by_category = df.groupby("category")["amount"].sum().to_dict()

    prompt = f"""
    You are an expert personal finance AI agent.
    Analyze the following monthly expense summary and provide tailored recommendations:

    Total Spent: ${total_spent:.2f}
    Category Breakdown: {json.dumps(summary_by_category)}

    Provide:
    1. A concise overview of spending habits.
    2. Top 2-3 areas where unnecessary spending can be reduced.
    3. Actionable financial tips for better budgeting.
    """

    # Use the global 'model' variable
    response = client_obj.models.generate_content(
        model=model, # Using the global 'model' variable
        contents=prompt,
        config=types.GenerateContentConfig(temperature=0.4),
    )

    return response.text


def run_expense_agent():
    # client and model are expected to be globally defined from previous cells.
    # SAMPLE_EXPENSES is also expected to be globally defined.

    print("--- 🤖 Running Expense Analysis Agent ---")

    # 1. Categorize Expenses
    print("\n[1/3] Categorizing raw expense entries...")
    # Pass the global 'client' object
    df = categorize_expenses(client, SAMPLE_EXPENSES)
    print("\nCategorized Expense Table:")
    print(df.to_string(index=False))

    # 2. Generate Chart
    print("\n[2/3] Generating spending chart...")
    chart_file = plot_expense_breakdown(df)
    print(f"📊 Visualization saved to '{chart_file}'\n")

    # 3. Analyze Habits & Provide Advice
    print("[3/3] Generating AI financial recommendations...")
    # Pass the global 'client' object
    advice = generate_financial_advice(client, df)

    print("\n--- 💡 Financial Insights & Suggestions ---")
    print(advice)


if __name__ == "__main__":
    run_expense_agent()
```
