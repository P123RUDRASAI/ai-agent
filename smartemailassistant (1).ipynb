```python
from google import genai

class EmailAgent:

    def __init__(self, api_key):
        self.client = genai.Client(api_key=api_key)

    def generate_email(self, recipient, purpose, tone, length):

        prompt = f"""
You are an expert email assistant.

Generate a complete email.

Recipient: {recipient}

Purpose: {purpose}

Tone: {tone}

Length: {length}

Return the response in this format:

Subject:
<subject>

Body:
<body>
"""

        response = self.client.models.generate_content(
            model="gemini-3.5-flash",   # or gemini-2.5-flash if needed
            contents=prompt
        )

        return response.text
```

```python
API_KEY = "replace with u r key"

agent = EmailAgent(API_KEY)

email = agent.generate_email(
    recipient="HR Manager",
    purpose="Request leave for two days due to fever",
    tone="Professional",
    length="Medium"
)

print(email)
```

```python
#building real ai agent
class PromptBuilder:

    @staticmethod
    def generate_email(recipient, purpose, tone, length):
        return f"""
You are an expert Email Assistant AI.

Recipient: {recipient}
Purpose: {purpose}
Tone: {tone}
Length: {length}

Write a professional email.

Return:

Subject:
...

Body:
...
"""
```

```python
prompt = PromptBuilder.generate_email(
    recipient="HR Manager",
    purpose="Request leave for two days due to fever",
    tone="Professional",
    length="Medium"
)

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)

print(response.text)
```
